```python
import pandas as pd
import matplotlib.pyplot as plt

import plot_benchmark
from pathlib import Path
import os
from loguru import logger

logger.remove()
os.chdir("..")

%load_ext autoreload
%autoreload
%matplotlib inline
```

```python
# Set working directory to parent dir


directory = Path("./cmake-build-release")
targets = plot_benchmark.find_files(directory, "vec_*")
files = plot_benchmark.run_benchmarks(targets)
df = plot_benchmark.load_reports(files)
df
```

```python
import seaborn as sns
```

```python
# Visualization: Line plot of 'size' vs. 'cpu_time' with 'framework' as color, scaled logarithmically
fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(x='Problem Size [N]', y='Runtime', hue='Framework', data=df, marker='o', ax=ax)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_ylabel('Runtime (nanoseconds)')
ax.set_title(f'Runtime in nanoseconds vs. Problem Size $N$ for {df["Problem"].unique()[0]}')
ax.grid(True, which='both', linestyle='--', alpha=0.7)
```
